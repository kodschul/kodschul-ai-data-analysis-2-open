## Umsatz 2025 je Region

Die Zelle filtert das Jahr 2025, bündelt die Zeilen nach `region` und summiert den Umsatz je Gruppe.

```python
umsatz_2025 = df[df["jahr"] == 2025].groupby("region")["umsatz"].sum().round(2)
umsatz_2025
```
